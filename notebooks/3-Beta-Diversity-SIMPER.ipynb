## Introduction and Scope

This analysis uses **Similarity Percentage (SIMPER)** to identify fungal taxa that contribute most strongly to observed differences in community composition among selected spatial and temporal groups. SIMPER decomposes average Bray–Curtis dissimilarity into taxon-level contributions, allowing the taxa accounting for the largest portions of between-group dissimilarity to be identified. The analyses presented here are designed as targeted follow-ups to the beta diversity results, which detected significant differences in fungal community composition among transects and significant temporal variation within each transect.

The scope includes three analytical components: **saproxylic fungi among Transects A, B, and C; and saproxylic fungi across Weeks Elapsed within each of Transects A, B, and C**. The transect analyses evaluate spatial differences through the pairwise contrasts A–B, A–C, and B–C, whereas the temporal analyses examine which taxa contribute most to compositional turnover across sampling periods within each transect. Because SIMPER is descriptive rather than a formal test of taxon–group association, identified taxa are interpreted as major contributors to observed dissimilarity and as candidates for further ecological investigation rather than as definitive indicator taxa.


### Basis for SIMPER Selection

The SIMPER comparisons were selected as targeted follow-up analyses to the beta diversity results rather than being applied indiscriminately to every environmental grouping. At the landscape scale, fungal community composition differed significantly among Transects A, B, and C for both the complete fungal community and the saproxylic fungal subset. For all fungi, transect explained 11.3% of compositional variation (PERMANOVA: p = 0.001, R² = 0.113), with significant separation also detected by ANOSIM (p = 0.001, R = 0.392). Similarly, saproxylic fungal composition differed among transects (PERMANOVA: p = 0.001, R² = 0.102; ANOSIM: p = 0.001, R = 0.228). These results provide a statistical basis for using SIMPER to determine which taxa contribute most to the pairwise dissimilarities among Transects A, B, and C for both fungal datasets (Table 1). **Because multivariate dispersion also differed among transects, however, SIMPER results are interpreted as contributions to average dissimilarity rather than as evidence that individual taxa independently drive centroid separation.**

Temporal SIMPER analyses were also selected for Weeks Elapsed within all three transects because temporal variation was consistently associated with saproxylic fungal community composition (Table 2). Weeks Elapsed was significant in Transect A (PERMANOVA: p = 0.001, R² = 0.091; ANOSIM: p = 0.001, R = 0.175), Transect B (PERMANOVA: p = 0.001, R² = 0.075; ANOSIM: p = 0.001, R = 0.374), and Transect C (PERMANOVA: p = 0.001, R² = 0.050; ANOSIM: p = 0.001, R = 0.241). These results indicate a consistent temporal signal across the study landscape and justify investigating which saproxylic taxa contribute to temporal compositional differences within each transect. Interpretation nevertheless requires different levels of caution: multivariate dispersion also differed significantly among Weeks Elapsed groups in Transects A (p = 0.005) and C (p = 0.002), whereas no significant dispersion difference was detected in Transect B (p = 0.066). Accordingly, SIMPER results for A and C may reflect both temporal shifts in composition and changes in within-period community heterogeneity, while Transect B provides comparatively cleaner evidence of temporal compositional turnover. Wood Texture and Wood Type were not selected for SIMPER because PERMANOVA provided little or no evidence of consistent compositional differentiation within the transects.

### Research Objectives

Based on the beta diversity patterns identified across spatial and temporal groupings, the objective of this SIMPER analysis is to determine which fungal taxa account for the largest proportions of the observed community dissimilarities. Specifically, the analysis addresses the following research questions:

1. **Spatial Differentiation — Saproxylic Fungi:** Which saproxylic fungal taxa contribute most to the average community dissimilarity among Transects A, B, and C, particularly for the pairwise comparisons A–B, A–C, and B–C?

2. **Temporal Differentiation — Saproxylic Fungi:** Which saproxylic fungal taxa contribute most to temporal differences in community composition across Weeks Elapsed within Transects A, B, and C?

3. **Cross-Transect Temporal Comparison:** Are the taxa contributing most strongly to temporal turnover shared among transects, or does each transect exhibit a distinct set of taxa associated with temporal community change?

Together, these analyses extend the community-level beta diversity findings by identifying the taxa contributing most strongly to observed **spatial differentiation and temporal turnover** in fungal assemblages.


## Key Findings

These results should be interpreted as identifying taxa that contribute strongly to average between-transect dissimilarity, rather than as evidence that the taxa are definitive indicators. Indicator Species Analysis can subsequently be used to evaluate whether the prominent SIMPER taxa also show strong and consistent associations.

### Contributors to Saproxylic Fungal Dissimilarity Among Transects

SIMPER analysis identified distinct sets of saproxylic fungal taxa contributing to pairwise Bray–Curtis dissimilarity among Transects A, B, and C (Table 3). **The A–B comparison was dominated by a small number of taxa, with Xenasmatella (20.2%), Hyaloscypha (16.7%), and Serpula (15.2%) together accounting for approximately 52.2% of cumulative dissimilarity.** All three were statistically significant contributors (p ≤ 0.002).

The A–C comparison showed a more distributed pattern of taxon contributions. **Hyaloscypha was the largest contributor (14.3%) and was significantly more abundant in Transect A (p = 0.025).** Other significant contributors included Phellinus (6.5%; higher in A), Scutellinia (5.1%; higher in C), Cistella (4.9%; higher in C), Arachnopeziza (4.6%; higher in C), and Ceraceomyces (4.5%; higher in C). This pattern suggests that the difference between Transects A and C reflects contributions from several taxa rather than being dominated by only one or two highly influential species.

For the B–C comparison, **Xenasmatella was again the largest contributor, accounting for 16.8% of average dissimilarity, and was substantially more abundant in Transect B (p = 0.006).** Several taxa contributed significantly in the opposite direction and were more abundant in Transect C, including Curvibasidium (4.2%, p = 0.002), Arachnopeziza (4.0%, p = 0.001), and Cistella (3.7%, p = 0.017). These results indicate that differences between B and C reflect both the strong representation of Xenasmatella in Transect B and a broader group of taxa occurring at higher abundance in Transect C.

Across the three pairwise transect contrasts, **Xenasmatella and Hyaloscypha emerged as the strongest recurring contributors to saproxylic fungal dissimilarity, each appearing in all three contrasts and showing significant contributions in two (Table 4).** Cistella and Arachnopeziza also showed consistent statistical support, with significant contributions in both contrasts in which they occurred, although their individual contribution percentages were smaller. In contrast, taxa such as Serpula and Phellinus contributed substantially to overall dissimilarity but were significant in only one of three contrasts, while Pragmopora was recurrent across all three contrasts but was not permutation-significant in any.

**Xenasmatella and Serpula had higher mean abundance in Transect B, whereas Hyaloscypha was more abundant in Transect A.** Entoleuca, although contributing a smaller proportion (4.5%), was also significantly associated with this contrast and was more abundant in Transect B. 

Several taxa also showed contrast-specific associations with Transect C. **Scutellinia, Cistella, Arachnopeziza, Ceraceomyces, and Curvibasidium were significant contributors in comparisons involving Transect C and had higher mean abundance there, indicating that they contributed to the compositional distinctiveness of the Transect C saproxylic fungal assemblage.**

Overall, the SIMPER results suggest that Transect B is differentiated partly by a few dominant taxa, whereas Transect C is characterized by a broader assemblage of taxa contributing to compositional dissimilarity.

### Temporal Contributors to Saproxylic Fungal Dissimilarity in Transect A

SIMPER analysis decomposed the temporal Bray–Curtis dissimilarities among Weeks Elapsed groups within Transect A and identified a relatively small set of taxa that repeatedly contributed substantially to temporal community differences (Table 5). **Across the pairwise comparisons, Xenasmatella, Hyaloscypha, Serpula, Metulodontia, Ceraceomyces, Lenzites, Collophora, and Pragmopora emerged as particularly important recurring contributors.** However, taxa with the largest numerical contributions were not always significant in permutation tests, demonstrating that a large contribution to dissimilarity does not necessarily indicate a consistent temporal association.

***Xenasmatella***

Xenasmatella emerged as the strongest recurring temporal contributor overall, occurring in 27 significant contrasts with a mean contribution of 36.6% and a maximum contribution of 46.5%. Its observed abundance peaked at Week 68, and several of its largest significant contributions involved late sampling periods, particularly Weeks 57, 67, and 68. For example, Xenasmatella contributed 35.9% of dissimilarity in the Week 10–57 contrast, 36.3% in Week 10–67, and 35.5% in Week 10–68. Although only 40.7% of its significant contrasts involved its exact peak at Week 68, the repeated high contributions associated with later sampling periods support a strong late-stage temporal pattern.

***Metulodontia, Serpula, Ceraceomyces, Lenzites, and Collophora***

Several taxa displayed substantially more concentrated taxon-specific temporal patterns. Metulodontia showed one of the clearest associations, with peak abundance at Week 65 and all 18 of its significant contrasts involving that period. Its mean contribution across significant contrasts was 23.5%, reaching a maximum of 35.1%. Serpula peaked at Week 8, which was involved in 16 of its 22 significant contrasts (72.7% peak support), with contributions averaging 23.2%.

Three additional taxa showed especially strong temporal concentration. Ceraceomyces peaked at Week 10, and all 17 of its significant contrasts involved that period; its contributions averaged 8.7% and reached 10.3%. Likewise, all 11 significant contrasts for Lenzites involved its Week 12 peak, while all 12 significant contrasts for Collophora involved its Week 14 peak. These patterns provide strong evidence that particular sampling periods were repeatedly associated with elevated contributions from specific saproxylic taxa.

***Pragmopora, Hyaloscypha, and Phellinus***

Pragmopora also displayed a localized early temporal pattern. Its abundance peaked at Week 2, and all six of its significant contrasts involved that peak period, with a mean contribution of 15.3% and a maximum of 16.7%.

In contrast, Hyaloscypha contributed strongly to temporal dissimilarity but showed less concentration around a single sampling period. It peaked at Week 4 and occurred in 12 significant contrasts, but only five involved its peak week (41.7% peak support). Nevertheless, its contribution magnitude was substantial, averaging 31.2% and reaching 44.4%. Phellinus likewise appeared frequently among taxa accounting for approximately 70% of cumulative dissimilarity, but only two of its contrasts were significant and neither involved its observed abundance peak at Week 10. These results illustrate the distinction between a taxon's numerical contribution to Bray–Curtis dissimilarity and the temporal consistency of that contribution.

**Sequence of Observed Peaks Among Major Recurring Contributors** 
When the observed abundance peaks of the major recurring contributors are arranged chronologically, the results suggest the following temporal pattern:

> Pragmopora (Week 2) → Hyaloscypha (Week 4) → Serpula (Week 8) → Ceraceomyces (Week 10) → Lenzites (Week 12) → Collophora (Week 14) → Metulodontia (Week 65) → Xenasmatella (Week 68)

The strength of peak support varies among these taxa. **The temporal associations of Pragmopora, Ceraceomyces, Lenzites, Collophora, and Metulodontia were tightly concentrated around their respective peak periods, whereas Hyaloscypha and Xenasmatella contributed significantly across a broader range of sampling periods.** The sequence should therefore be interpreted as an observed ordering of abundance peaks among major SIMPER contributors, rather than as evidence of ecological succession.

Overall, the temporal SIMPER results indicate a dynamic and taxon-specific pattern of community turnover within Transect A. **Early and intermediate sampling periods were differentiated by a sequence of taxa including Pragmopora, Hyaloscypha, Serpula, Ceraceomyces, Lenzites, and Collophora, while the latest periods were distinguished particularly by Metulodontia at Week 65 and Xenasmatella toward Weeks 67–68.** These taxa provide strong candidates for comparison with temporal SIMPER patterns in Transects B and C and for subsequent evaluation using Indicator Species Analysis or other taxon-specific temporal analyses.

### Temporal Contributors to Saproxylic Fungal Dissimilarity in Transect B

SIMPER analysis revealed pronounced temporal variation in saproxylic fungal community composition within Transect B, with a recurring group of taxa accounting for substantial proportions of Bray–Curtis dissimilarity among Weeks Elapsed groups (Table 6). **The most prominent recurring contributors included Xenasmatella, Serpula, Coniophora, Phellinus, Hyaloscypha, Curvibasidium, Entoleuca, and Sistotremastrum.** Several additional taxa, including Pragmopora, Skvortzovia, and Ascocoryne, showed more temporally localized patterns. As in Transect A, large numerical SIMPER contributions were not always permutation-significant, emphasizing the distinction between contribution magnitude and the consistency of that contribution across samples.

***Xenasmatella and Serpula***

**Xenasmatella was the most frequently significant contributor, occurring in 38 significant contrasts with a mean contribution of 33.0% and a maximum contribution of 45.3%.** Its observed mean abundance peaked at Week 51 (0.984), but only 18.4% of its significant contrasts involved that exact peak period. This indicates that Xenasmatella contributed strongly to temporal dissimilarity across a broad range of sampling periods rather than being tightly associated with a single temporal peak. For example, it made large significant contributions in contrasts involving Weeks 51, 55, 59, 61, and 63, including 39.8% in Week 51–52, 45.3% in Week 14–51, and 42.7% in Week 14–55.

**Serpula was similarly influential, occurring in 25 significant contrasts with an average contribution of 36.3% and a maximum of 47.0%.** Its abundance peaked at Week 14 (0.981), although only 48.0% of significant contrasts involved the peak week. Nevertheless, several of the strongest pairwise differences centered on Week 14: Serpula contributed 41.8% to Week 14–16, 45.4% to Week 14–51, 47.0% to Week 14–55, and 46.4% to Week 14–59, with each of these contrasts permutation-significant. The results therefore support a strong Week 14 signal for Serpula, while also indicating that its influence extended beyond that single sampling period.

***Curvibasidium, Phellinus, and Entoleuca***

Several taxa exhibited clearer early temporal localization. **Curvibasidium peaked at Week 4** with a mean abundance of 0.335, and all 17 of its significant contrasts involved Week 4, providing 100% peak support. Its significant contributions averaged 13.7% and reached 18.6%. **Phellinus also peaked at Week 4 (0.296)** and occurred in 20 significant contrasts, 16 of which involved its peak period (80.0% peak support). Together, these results identify Week 4 as an important early sampling period characterized by elevated contributions from both taxa.

**Entoleuca showed another strongly localized early pattern. Its abundance peaked at Week 6 (0.501), and 12 of its 14 significant contrasts** involved that period, corresponding to 85.7% peak support. Its contributions averaged 20.1% and reached 25.7%. Significant contrasts involving Week 6 repeatedly showed higher abundance of Entoleuca at that period, including comparisons with Weeks 12, 14, 16, 51, 52, 53, 55, 59, and 61.

***Hyaloscypha, Coniophora, and Sistotremastrum***

The later sampling periods were characterized by a different set of recurring taxa. **Hyaloscypha showed one of the strongest late temporal signals, peaking at Week 67 with a mean abundance of 0.895.** It occurred in 18 significant contrasts, 16 of which involved Week 67, yielding 88.9% peak support. Its contributions were also large, averaging 36.9% and reaching a maximum of 51.7%. Significant contrasts consistently showed greater abundance at Week 67 relative to many earlier periods, making Hyaloscypha one of the clearest late-stage temporal contributors in Transect B.

**Two taxa were especially prominent at Week 68. Sistotremastrum peaked at Week 68 (0.419), with 13 of its 14 significant contrasts involving that period (92.9% peak support). Coniophora also peaked at Week 68, reaching a substantially higher mean abundance of 0.744.** It was significant in 32 contrasts and had a mean contribution of 27.2%, with a maximum of 34.8%. Although only 53.1% of its significant contrasts involved its exact peak, Coniophora repeatedly distinguished Week 68 from earlier sampling periods and also contributed significantly to contrasts involving Week 65. The Week 67–68 contrast illustrates this late turnover particularly clearly: Coniophora contributed 34.8% and was higher at Week 68, whereas Hyaloscypha contributed 33.1% and was higher at Week 67.

***Other Temporally Localized Contributors***

Several secondary contributors exhibited highly concentrated temporal associations despite occurring in fewer significant contrasts. Skvortzovia peaked at Week 52, and all five of its significant contrasts involved that period. Ascocoryne similarly peaked at Week 61, with all five significant contrasts involving its peak. Pragmopora peaked at Week 8 and occurred in ten significant contrasts, although only four involved that peak period, indicating a less temporally concentrated contribution. Phanerochaete and Panellus each showed a significant contribution associated with their Week 2 abundance peak, but each occurred in only one significant contrast and therefore represents a more isolated temporal signal.

***Sequence of Observed Peaks Among Major Recurring Contributors***

When the observed abundance peaks of the principal recurring contributors are ordered chronologically, Transect B shows the following temporal pattern:

> Curvibasidium / Phellinus (Week 4) → Entoleuca (Week 6) → Pragmopora (Week 8) → Serpula (Week 14) → Xenasmatella (Week 51) → Skvortzovia (Week 52) → Ascocoryne (Week 61) → Hyaloscypha (Week 67) → Sistotremastrum / Coniophora (Week 68)

The strength of support for these peak associations varied considerably. **Curvibasidium, Entoleuca, Skvortzovia, Ascocoryne, Hyaloscypha, and Sistotremastrum showed strong concentration of significant contrasts around their respective peak periods. In contrast, Pragmopora, Serpula, Xenasmatella, and Coniophora contributed significantly across a broader range of sampling periods, indicating that their influence on temporal beta diversity was not restricted to their exact abundance maxima.** 

**Accordingly, this pattern should be interpreted as a chronological sequence of observed abundance peaks among major SIMPER contributors, rather than as direct evidence of fungal succession.**

Overall, the temporal SIMPER results indicate that saproxylic fungal turnover in Transect B involved distinct early, intermediate, and late taxon-specific patterns. Early temporal differentiation was associated particularly with Curvibasidium, Phellinus, and Entoleuca; Week 14 was strongly characterized by Serpula; the intermediate-to-late period featured substantial contributions from Xenasmatella; and the latest sampling periods were increasingly differentiated by Hyaloscypha, Sistotremastrum, and Coniophora. These recurring and temporally localized taxa provide strong candidates for comparison with the corresponding patterns in Transects A and C and for subsequent evaluation using Indicator Species Analysis or other taxon-specific temporal analyses.

### Temporal Contributors to Saproxylic Fungal Dissimilarity in Transect C

SIMPER analysis revealed clear temporal variation in saproxylic fungal community composition within Transect C, with a series of taxa showing strongly localized contributions across Weeks Elapsed groups (Table 7). **The most prominent recurring contributors included Curvibasidium, Serpula, Panellus, Phanerochaete, Ceraceomyces, and Hyaloscypha, while several other taxa—including Xenasmatella, Collophora, Pragmopora, Coniophora, Corticium, and Oliveonia—showed more restricted temporal associations.** In contrast with Transects A and B, many of the significant contributors in Transect C showed very high peak support, indicating that their significant SIMPER contrasts were often tightly centered on the week of maximum observed abundance.

***Curvibasidium and Early Temporal Differentiation***

**Curvibasidium showed the clearest early temporal signal, peaking at Week 2 with a mean abundance of 0.516.** All seven of its significant contrasts involved this peak period, corresponding to 100% peak support. Its contribution averaged 18.5% across significant contrasts and reached 20.4%. Significant contributions in contrasts between Week 2 and Weeks 4, 6, 8, 12, 14, 16, and 52, consistently reflecting greater abundance at Week 2. This indicates that **Curvibasidium was a major component of early community differentiation in Transect C.**

Several additional taxa contributed to early temporal differences. Phellinus, Trametes, and Serpula all peaked at Week 4. Although Phellinus and Trametes each occurred in only one significant contrast, Serpula was significant in six contrasts, all involving its Week 4 peak. Its contributions averaged 10.5% and reached 12.1%, indicating a more recurrent early-period association. Collophora subsequently peaked at Week 6, with both of its significant contrasts involving that sampling period.

***Panellus and Phanerochaete***

**A particularly strong mid-early temporal pattern occurred at Week 10. Panellus reached its maximum mean abundance at this period (0.697) and was significant in seven contrasts, all involving Week 10.** Its contributions averaged 23.4% and reached 25.6%, making it one of the strongest recurring contributors in Transect C. Significant contributions in comparisons between Week 10 and Weeks 4, 6, 8, 12, 14, 16, and 52.

**Phanerochaete displayed an almost identical pattern. It peaked at Week 10 with a mean abundance of 0.524, and all seven of its significant contrasts involved that period.** Its mean contribution was 17.1%, with a maximum of 18.9%. The consistently high Average/SD values associated with these contrasts further indicate that its contribution was relatively stable across pairwise sample comparisons. Together, Panellus and Phanerochaete define one of the clearest taxon-specific temporal signatures in Transect C.

***Ceraceomyces and Week 14***

**Ceraceomyces showed another strongly localized pattern centered on Week 14, where its mean abundance peaked at 0.445.** All six of its significant SIMPER contrasts involved this period, yielding 100% peak support. Its contribution averaged 18.9% and reached 20.2%. Significant contrasts included Week 14 against Weeks 4, 6, 8, 10, 12, 16, and 52, with Week 14 consistently showing higher abundance where significant. This pattern identifies Ceraceomyces as a strong contributor to community differentiation during this sampling period.

**Several other taxa also showed associations around Weeks 12–14. Pragmopora, Coniophora, and Corticium peaked at Week 12, while Ascocoryne peaked at Week 14. However, each of these taxa occurred in relatively few significant contrasts compared with Ceraceomyces, suggesting more limited temporal influence.**

***Hyaloscypha and the Later Sampling Period***

**The strongest later temporal signal was associated with Hyaloscypha, which peaked at Week 16 with a mean abundance of 0.717.** It appeared in 11 significant contrasts and had a mean contribution of 23.8%, reaching a maximum of 26.6%. However, only six of the 11 significant contrasts involved its exact peak week, corresponding to 54.5% peak support. This indicates that, although Week 16 represented its abundance maximum, Hyaloscypha contributed to temporal dissimilarity across a broader range of comparisons than taxa such as Panellus, Phanerochaete, or Ceraceomyces. Significant contrasts involving Week 16 included comparisons with Weeks 2, 4, 6, 8, 12, and 14.

A later and more isolated signal was observed for Oliveonia, which peaked at Week 52 with a mean abundance of 0.245. Its single significant contrast involved this peak period, indicating a localized but comparatively weak contribution relative to the major recurring taxa.

***Sequence of Observed Peaks Among Major Recurring Contributors***

When the observed abundance peaks of the major recurring contributors are arranged chronologically, Transect C shows the following temporal pattern:

> Curvibasidium (Week 2) → Serpula (Week 4) → Collophora (Week 6) → Panellus / Phanerochaete (Week 10) → Ceraceomyces (Week 14) → Hyaloscypha (Week 16) → Oliveonia (Week 52)

**The strongest evidence for temporally localized peaks was observed for Curvibasidium, Serpula, Panellus, Phanerochaete, and Ceraceomyces, each of which showed 100% peak support across its significant contrasts. Hyaloscypha contributed strongly but was less tightly centered on its abundance maximum, while Oliveonia represented a more isolated late-period signal.**

Overall, the temporal SIMPER results indicate that saproxylic fungal turnover in Transect C was characterized by a relatively compressed sequence of strongly localized taxon-specific peaks during the earlier sampling period, followed by fewer prominent changes later in the study. Early community differentiation was associated with Curvibasidium at Week 2 and Serpula at Week 4, followed by strong Week 10 signals from Panellus and Phanerochaete, a Week 14 peak in Ceraceomyces, and a Week 16 peak in Hyaloscypha. Compared with Transects A and B, Transect C therefore appears to show a more concentrated early-to-intermediate temporal turnover pattern, with fewer strongly recurring taxa peaking during the latest sampling periods. These patterns provide a useful basis for cross-transect comparison and for subsequent evaluation using Indicator Species Analysis or other taxon-specific temporal models.

### Cross-Transect Temporal Comparison

The cross-transect comparison revealed a **shared core of temporal SIMPER contributors, but substantial differences in when those taxa reached their maximum observed abundance**. Of the 24 taxa identified as temporal contributors across the three transects, eight—*Collophora*, *Hyaloscypha*, *Panellus*, *Phanerochaete*, *Phellinus*, *Pragmopora*, *Serpula*, and *Xenasmatella*—occurred as contributors in all three transects (Table 8). Five additional taxa occurred in two transects, whereas 11 were transect-specific. Thus, temporal community dissimilarity was not driven entirely by different taxa among transects; instead, a considerable portion involved the **same recurring taxa exhibiting different temporal patterns**.

The clearest cross-transect distinction was the **asynchrony in peak timing among these recurring taxa** (Table 8; Figure 1). *Xenasmatella* showed the strongest example, reaching its observed abundance peak at Week 68 in Transect A, Week 51 in Transect B, but only Week 2 in Transect C. *Hyaloscypha* likewise peaked at markedly different periods—Week 4 in A, Week 67 in B, and Week 16 in C. Other shared taxa displayed smaller but still distinct shifts: *Serpula* peaked at Weeks 8, 14, and 4 in A, B, and C, respectively, while *Pragmopora* peaked progressively later from A (Week 2) to B (Week 8) to C (Week 12). These patterns indicate that recurrence of a taxon across transects did **not imply synchronized temporal abundance dynamics**.

Some taxa nevertheless showed partial temporal agreement between transects. *Panellus* and *Phanerochaete* both peaked at Week 2 in Transects A and B before shifting to Week 10 in C, while *Phellinus* peaked at Week 4 in both B and C but at Week 10 in A. *Collophora* also showed relatively early peaks in all three transects, although its maximum shifted from Week 4 in B to Week 6 in C and Week 14 in A. These shared and shifted peaks suggest that the temporal behavior of recurring contributors ranged from relatively conserved to strongly transect-dependent. 

The transects also differed in the overall distribution of peak timing. **Transect A exhibited the broadest temporal spread**, with peaks extending from Week 2 through Week 68. Most contributors peaked during Weeks 2–16, but several prominent taxa occurred much later, including *Arachnopeziza* at Week 51, *Metulodontia* at Week 65, and *Xenasmatella* at Week 68. Among the strongest mean SIMPER contributors in A were *Xenasmatella* (36.3%), *Hyaloscypha* (31.2%), *Metulodontia* (23.5%), and *Serpula* (23.2%). 

**Transect B showed a particularly pronounced separation between early- and late-peaking contributors.** Early peaks occurred from Weeks 2–14 for taxa such as *Curvibasidium*, *Phellinus*, *Entoleuca*, *Pragmopora*, and *Serpula*, whereas a second group peaked from Weeks 51–68, including *Xenasmatella*, *Skvortzovia*, *Ascocoryne*, *Hyaloscypha*, *Sistotremastrum*, and *Coniophora*. Several of the largest mean contributors belonged to this temporal structure, including *Serpula* (36.3%), *Hyaloscypha* (35.8%), *Xenasmatella* (33.0%), and *Coniophora* (27.2%). 

In contrast, **Transect C was dominated by early abundance peaks**. Nearly all temporal contributors reached their maxima between Weeks 2 and 16; *Oliveonia*, which peaked at Week 52, was the major exception. Notable contributors included *Panellus* (23.4%; Week 10), *Hyaloscypha* (23.8%; Week 16), *Ceraceomyces* (18.9%; Week 14), and *Curvibasidium* (18.5%; Week 2). Particularly striking was *Xenasmatella*: whereas it was a late-peaking and high-contribution taxon in A and B, it peaked at Week 2 in C and had a considerably smaller mean contribution of 8.0%. 

Figure X also shows that **a large mean SIMPER contribution did not necessarily coincide with strong support for the observed abundance peak**. For example, *Xenasmatella* had the highest mean contribution in Transect A (36.3%) but only 36.7% of its significant contrasts involved its Week 68 peak; in Transect B, its mean contribution was similarly high (33.0%) but peak support at Week 51 was only 16.7%. By comparison, its Week 2 peak in Transect C had 100% peak support despite a much smaller mean contribution. A similar distinction is visible for other taxa, indicating that **overall importance to temporal dissimilarity and the degree to which that dissimilarity is specifically associated with the observed peak are complementary, rather than interchangeable, measures**. 

Overall, Tables X–Y and Figure X indicate that saproxylic fungal temporal dynamics across the Oulanka transects were characterized by a **combination of taxonomic recurrence and temporal turnover**. A common set of taxa repeatedly contributed to temporal Bray–Curtis dissimilarity, but their peak timing, magnitude of contribution, and peak support frequently differed among transects. The strongest cross-transect signal is therefore not simply the presence of different temporal contributors, but the **transect-specific timing and expression of many of the same recurring fungal taxa**.

### Tasks

1. Identify the taxa that are most important to the SIMPER dissimilarities.
2. Identify whether particular taxa show **temporal peaks**.

```
SIMPER results
     │
     ├── Which taxa explain most dissimilarity?
     │       ↓
     │   cumulative contribution ≈ 70%
     │
     ├── Which contributions are statistically supported?
     │       ↓
     │   p ≤ 0.05
     │
     ├── Which taxa recur across contrasts?
     │       ↓
     │   count significant contrasts
     │
     ├── Are their contributions consistent?
     │       ↓
     │   examine Average / SD
     │
     └── Important SIMPER taxa
               │
               ↓
       Reconstruct abundance
       from Group_1/ava and
       Group_2/avb
               │
               ↓
       Mean abundance by
       Taxon × Week
               │
               ↓
       Find maximum abundance
       for each taxon
               │
               ↓
       Order peaks by week
               │
               ↓
       Verify peak weeks occur
       in significant SIMPER
       contrasts
```

### Definitions

* **Important SIMPER contributor:** a taxon accounting for a substantial portion of between-group dissimilarity, especially when its contribution is permutation-significant, consistent, and recurrent across contrasts.

* **Taxon-specific peak:** the Week Elapsed at which that taxon has its highest observed mean abundance.

* **Supported temporal peak:** a taxon-specific abundance peak that is also repeatedly involved in significant SIMPER contrasts.


***Only the last two steps give you the basis for saying:***
> “The data show a temporal sequence of observed abundance peaks.”

That is not consider as that sequence ecological succession yet. A temporal abundance plot plus Indicator Species Analysis or a dedicated temporal model would provide stronger evidence for that interpretation.

#### run_simper

````r
run_simper <- function(ordination, 
                       majority_threshold = 0.70,
                       pvalue_threshold = 0.05) {
    ordination_configuration <- ordination$configuration
    # 0. Similarity Percentages: identify which species/taxa drives the differences between groups
    simper <- simper(ordination_configuration$community_hellinger, 
                     group = ordination_configuration$environment[[
                         ordination_configuration$variable
                     ]], 
                     permutations = 999)
    simper_summary <- summary(simper)
    simper_result_df <- as.data.frame(simper_summary)

    
    # 1. Retain the major SIMPER contributors
    # Which taxa account for the major portion (~70%) of dissimilarity in each contrast?
    major_contributors_df <- simper_result_df %>%
        group_by(Contrast) %>%
        arrange(cumsum, .by_group = TRUE) %>%
        mutate(
            row_id = row_number(),
            cutoff_row = {
                x <- which(cumsum >= majority_threshold)
    
                if (length(x) == 0) {
                    n()
                } else {
                    x[1]
                }
            }
        ) %>%
        filter(row_id <= cutoff_row) %>%
        ungroup()

    
    # 2. Identify permutation-significant contributors
    # Which of those major contributors are statistically significant at p ≤ 0.05?
    significant_contributors_df <- major_contributors_df %>%
        filter(p <= pvalue_threshold)

    
    # 3a. Within each contrast, which taxa contribute the most, regardless of significance?
    contributors_by_contrast_df <- major_contributors_df %>%
        arrange(
            Contrast,
            desc(Contribution_pct)
        )

    
    # 3b. Rank important taxa within each contrast
    # Within each contrast, which significant taxa contribute the most?
    significant_by_contrast_df <- significant_contributors_df %>%
        arrange(
            Contrast,
            desc(Contribution_pct)
        )

    
    # 4. Determine which taxa recur across many significant contrasts
    # Which taxa repeatedly appear as significant contributors across many contrasts?
    taxon_recurrence_df <- significant_contributors_df %>%
        group_by(Taxon) %>%
        summarise(
            Significant_contrasts = n_distinct(Contrast),
            Mean_contribution_pct = mean(Contribution_pct),
            Median_contribution_pct = median(Contribution_pct),
            Max_contribution_pct = max(Contribution_pct),
            Mean_ratio = mean(ratio, na.rm = TRUE),
            .groups = "drop"
        ) %>%
        arrange(
            desc(Significant_contrasts),
            desc(Mean_contribution_pct)
        )

    
    # 5. Incorporate contribution consistency (ratio)
    # Which individual significant SIMPER results also have relatively consistent contributions, 
    # based on Average/SD?
    strong_contributors_df <- significant_contributors_df %>%
        filter(ratio >= 1) %>%
        arrange(
            desc(Contribution_pct),
            desc(ratio)
        )

    
    # 6. Create an overall contributor summary
    # Across the entire analysis, which taxa look most important when recurrence, 
    # contribution magnitude, and consistency are considered together?
    important_taxa_df <- major_contributors_df %>%
        group_by(Taxon) %>%
        summarise(
            Contrasts = n_distinct(Contrast),
            Significant_contrasts = n_distinct(
                Contrast[p <= pvalue_threshold]
            ),
            Significant_frequency = paste0(
                Significant_contrasts,
                " / ",
                Contrasts
            ),
            Higher_in_significant = paste(
                sort(unique(Higher_in[p <= pvalue_threshold])),
                collapse = ", "
            ),Mean_contribution = mean(Contribution_pct),
            Max_contribution = max(Contribution_pct),
            Median_ratio = median(ratio, na.rm = TRUE),
            .groups = "drop"
        ) %>%
        mutate(
            Higher_in_significant = if_else(
                Significant_contrasts == 0,
                "—",
                Higher_in_significant
            )
        ) %>%
        arrange(
            desc(Significant_contrasts),
            desc(Max_contribution)
        )

    
    return (list(
        summary = simper_summary,
        result_df = simper_result_df,
        major_contributors_df = major_contributors_df,
        significant_contributors_df = significant_contributors_df,
        contributors_by_contrast_df = contributors_by_contrast_df,
        significant_by_contrast_df = significant_by_contrast_df,
        taxon_recurrence_df = taxon_recurrence_df,
        strong_contributors_df = strong_contributors_df,
        important_taxa_df = important_taxa_df
    ))
}
````

#### identify_taxon_specific_abundance_peaks

````r
identify_taxon_specific_abundance_peaks <- function(simper) {
    simper_df <- simper$result_df
    significant_contributors_df <- simper$significant_contributors_df
    
    # Step 1-2: Extract abundances
    abundance_group1 <- simper_df %>%
        transmute(
            Taxon,
            Week = Group_1,
            Mean_abundance = ava
        )
    
    abundance_group2 <- simper_df %>%
        transmute(
            Taxon,
            Week = Group_2,
            Mean_abundance = avb
        )

    
    # Step 3: Combine abundances
    taxon_week_abundance <- bind_rows(
        abundance_group1,
        abundance_group2
    )

    
    # Step 4: Remove duplicated taxon-week abundance observations
    taxon_week_abundance <- taxon_week_abundance %>%
        distinct(
            Taxon,
            Week,
            Mean_abundance
        )

    
    # Step 4a: Verify that each Taxon x Week has only one abundance value
    # Result should be zero rows.
    #
    week_abundance_with_duplicates <- taxon_week_abundance %>%
        group_by(Taxon, Week) %>%
        summarise(
            n_values = n_distinct(Mean_abundance),
            .groups = "drop"
        ) %>%
        filter(n_values > 1)

    if (nrow(week_abundance_with_duplicates) > 0) {
        stop("Duplicate group abundance found")
    }

    
    # Step 4b: Collapse to get the actual abundance-by-time dataset.
    taxon_week_abundance <- taxon_week_abundance %>%
        group_by(Taxon, Week) %>%
        summarise(
            Mean_abundance = first(Mean_abundance),
            .groups = "drop"
        )

    
    # 5. Identify the peak week for each taxon
    # Answers the question: "At which group was each taxon's mean abundance highest?"
    taxon_peaks <- taxon_week_abundance %>%
        group_by(Taxon) %>%
        slice_max(
            order_by = Mean_abundance,
            n = 1,
            with_ties = TRUE
        ) %>%
        ungroup()

    
    # 6. Sort the peaks chronologically
    # The observed sequence of taxon-specific abundance peaks.
    # The basis for describing a sequence of observed abundance peaks.
    taxon_peaks <- taxon_peaks %>%
        mutate(
            Week = as.numeric(as.character(Week))
        ) %>%
        arrange(Week)

    
    # 7. Restrict the peak analysis to important SIMPER taxa
    # retain taxa that were significant SIMPER contributors at the transect level at least once
    candidate_taxa <- significant_contributors_df %>%
        distinct(Taxon)

    # Answer the question: "Among taxa that significantly contributed to temporal community dissimilarity, 
    #                       when did their observed abundance peaks occur?"
    # (RETURN)
    important_taxon_peaks_df <- taxon_peaks %>%
        semi_join(
            candidate_taxa,
            by = "Taxon"
        ) %>%
        arrange(Week)

    
    # 8. Require a taxon to be significant in, say, at least two contrasts before emphasizing its peak.
    recurring_taxa <- significant_contributors_df %>%
        count(
            Taxon,
            name = "Significant_contrasts"
        ) %>%
        filter(Significant_contrasts >= 2)
    
    # Taxa repeatedly implicated by SIMPER, ordered by their observed peak abundance week.
    # (RETURN)
    recurring_taxon_peaks_df <- taxon_peaks %>%
        inner_join(
            recurring_taxa,
            by = "Taxon"
        ) %>%
        arrange(Week)


    # 9. Verify the peak week actually participates in significant contrasts
    # Long representation of the significant contrasts
    significant_peak_support <- significant_contributors_df %>%
        select(
            Taxon,
            Contrast,
            Group_1,
            Group_2,
            Contribution_pct,
            ratio,
            p
        ) %>%
        pivot_longer(
            cols = c(Group_1, Group_2),
            names_to = "Group_position",
            values_to = "Week"
        )

    # Join with peak weeks
    peak_validation <- significant_peak_support %>%
        inner_join(
            taxon_peaks %>%
                select(
                    Taxon,
                    Peak_week = Week,
                    Peak_mean_abundance = Mean_abundance
                ),
            by = "Taxon"
        ) %>%
        mutate(
            Involves_peak = Week == Peak_week
        )

    # Summarize
    # (RETURN)
    peak_support_summary_df <- peak_validation %>%
        group_by(
            Taxon, Peak_week, Peak_mean_abundance) %>%
        summarise(
            Significant_contrasts = n_distinct(Contrast),
            Peak_related_contrasts =
                n_distinct(Contrast[Involves_peak]),
            Peak_support_pct =
                100 * Peak_related_contrasts /
                Significant_contrasts,
            .groups = "drop"
        ) %>%
        arrange(
            Peak_week,
            desc(Peak_support_pct)
        )

    return (list(
        important_taxon_peaks_df = important_taxon_peaks_df,
        recurring_taxon_peaks_df = recurring_taxon_peaks_df,
        peak_support_summary_df = peak_support_summary_df
    ))
}
````

#### create_simper_table

````r
create_simper_table <- function(major_contributors_df) {
    return (
        major_contributors_df %>%
            mutate(
                `Mean abundance` = paste0(
                    Group_1, ": ", sprintf("%.3f", ava),
                    " | ",
                    Group_2, ": ", sprintf("%.3f", avb)
                )
            ) %>%
            select(
                Contrast,
                Taxon,
                Contribution_pct,
                Cumulative_pct,
                `Mean abundance`,
                Higher_in,
                ratio,
                p,
                Significant
            )
    )
}
````

#### render_simper_table

````r
render_simper_table <- function(simper_table_df, caption_text) {
    simper_table <- simper_table_df %>%
        gt(groupname_col = "Contrast") %>%
        cols_label(
            Taxon = md("**Taxon**"),
            Contribution_pct = md("**Contribution (%)**"),
            Cumulative_pct = md("**Cumulative (%)**"),
            `Mean abundance` = md("**Mean abundance**"),
            Higher_in = md("**Higher in**"),
            ratio = md("**Average / SD**"),
            p = md("***p*-value**")
        ) %>%
        # Percentages
        fmt_number(
            columns = c(
                Contribution_pct,
                Cumulative_pct
            ),
            decimals = 1
        ) %>%
        # SIMPER ratio
        fmt_number(columns = ratio, decimals = 2) %>%
        # p-values
        fmt_number(columns = p, decimals = 3) %>%
        # Italicize fungal taxon names
        tab_style(
            style = cell_text(style = "italic"),
            locations = cells_body(columns = Taxon)
        ) %>%
        # Highlight statistically significant SIMPER results
        tab_style(
            style = cell_text(weight = "bold"),
            locations = cells_body(columns = p, rows = Significant)
        ) %>%   
        # Alignment
        cols_align(
            align = "left",
            columns = c(Taxon, `Mean abundance`)
        ) %>%
        cols_align(
            align = "center",
            columns = c(
                Contribution_pct,
                Cumulative_pct,
                Higher_in,
                ratio,
                p
            )
        ) %>%
        # Hide helper variable
        cols_hide(columns = Significant) %>%
        # Caption
        tab_caption(
            caption = md(caption_text)
        )

    simper_table %>%
        as_raw_html(inline_css = TRUE) %>%
        display_html()
}
````

#### create_temporal_contributor_summary_table

````r
create_temporal_contributor_summary_table <- function(simper, abundance_peaks) {
    significant_contributors_df <- simper$significant_contributors_df
    peak_support_summary_df <- abundance_peaks$peak_support_summary_df

    # Create the contributor summary
    contributor_summary_df <- significant_contributors_df %>%
        group_by(Taxon) %>%
        summarise(
            Total_contrasts = n_distinct(Contrast),
    
            Significant_contrasts_overall =
                n_distinct(Contrast[p <= 0.05]),
    
            Mean_contribution =
                mean(Contribution_pct),
    
            Max_contribution =
                max(Contribution_pct),
    
            Median_ratio =
                median(ratio, na.rm = TRUE),
    
            .groups = "drop"
        )
    
    
    # Create combined temporal contributor summary
    temporal_contributor_summary <- peak_support_summary_df %>%
        left_join(
            contributor_summary_df,
            by = "Taxon"
        ) %>%
        mutate(
            Significant_frequency = paste0(
                Significant_contrasts_overall,
                " / ",
                Total_contrasts
            )
        ) %>%
        select(
            Taxon,
            Peak_week,
            Peak_mean_abundance,
            Significant_contrasts_overall,
            Total_contrasts,
            Significant_frequency,
            Peak_related_contrasts,
            Peak_support_pct,
            Mean_contribution,
            Max_contribution,
            Median_ratio
        ) %>%
        arrange(
            as.numeric(as.character(Peak_week)),
            desc(Peak_support_pct),
            desc(Mean_contribution)
        )

    return (temporal_contributor_summary)
}
````

## Beta Diversity - SIMPER Analysis Notes

There are really **two different tasks** here:

1. identifying the taxa that are most important to the SIMPER dissimilarities; and
2. identifying whether particular taxa show **temporal peaks**.

Those should not be determined in exactly the same way.

### 1. How to identify important SIMPER contributors

I would use a hierarchy rather than relying on one column alone.

1. **Start with contribution percentage.** For each pairwise contrast, rank taxa by `Contribution_pct`. This tells you how much of the average Bray–Curtis dissimilarity is attributable to each taxon. A taxon contributing 25% is more influential to that particular contrast than one contributing 5%.

2. **Apply the cumulative contribution cutoff.** Retain taxa until cumulative contribution reaches about **70%**. This identifies the relatively small subset accounting for most of the observed dissimilarity. Your current tables already do this.

```
# Keeps the taxon that actually pushes the cumulative total past 70%.
simper_70 <- simper_df %>%
    filter(row_id <= cutoff_row)

## If cutoff_row is not already calculated, you can derive it from contrast:
simper_70 <- simper_df %>%
    group_by(Contrast) %>%
    arrange(row_id, .by_group = TRUE) %>%
    mutate(
        cutoff_row = which(Cumulative_pct >= 70)[1]
    ) %>%
    filter(row_id <= cutoff_row) %>%
    ungroup()
```

3. **Examine permutation significance.** Within that 70% subset, identify taxa with *p* ≤ 0.05. These are especially useful because their contribution appears more consistent than expected under permutation. I would call these **significant SIMPER contributors**, while taxa with high contribution but *p* > 0.05 remain important descriptively.

```
# Identify permutation-significant contributors
# Answers the question: Which taxa within the major ~70% SIMPER contributors also show significant permutation evidence?
significant_contributors <- simper_70 %>%
    filter(p <= 0.05)
```

4. **Look at `Average / SD`.** This is a measure of contribution consistency. Larger values indicate that the taxon's contribution is relatively consistent across the underlying sample-to-sample comparisons. A taxon with high contribution, significant *p*, and relatively high Average/SD is particularly compelling.

```
# Rank important taxa within each contrast
contributors_by_contrast <- simper_70 %>%
    arrange(
        Contrast,
        desc(Contribution_pct)
    ) %>%
    select(
        Contrast,
        Taxon,
        Contribution_pct,
        Cumulative_pct,
        ratio,
        p,
        Higher_in
    )

# Using the significant subset.
# This is the level you use when saying:
# "Xenasmatella contributed 36.3% of dissimilarity in the 10–67 contrast."
significant_by_contrast <- significant_contributors %>%
    arrange(
        Contrast,
        desc(Contribution_pct)
    )
```

5. **Use mean abundance and `Higher in` to interpret direction.** SIMPER tells you that a taxon contributes to a difference; the abundance columns tell you *why*. For example, if *Xenasmatella* contributes 36% to Week 10–67 and has mean abundance 0.000 at Week 10 versus 0.945 at Week 67, the difference is clearly associated with much greater abundance at Week 67.

6. **Look for recurrence across contrasts.** A taxon appearing as a significant contributor in many contrasts is more interesting than one appearing significantly in only one isolated comparison. This is how taxa such as *Xenasmatella*, *Metulodontia*, *Serpula*, and *Ceraceomyces* begin to stand out in your Transect A results.

### 2. How to identify taxon-specific temporal peaks

This requires a different approach. Strictly speaking, **SIMPER does not identify peaks**. It identifies taxa contributing to pairwise dissimilarities.

The most defensible way to identify a temporal peak is from the **mean abundance of each taxon at each Week Elapsed**, not from `Contribution_pct`.

Suppose you reconstruct this for *Xenasmatella*:

| Week | Mean abundance |
| ---: | -------------: |
|   10 |          0.000 |
|   52 |          0.507 |
|   57 |          0.908 |
|   63 |          0.664 |
|   65 |          0.414 |
|   67 |          0.945 |
|   68 |      **1.000** |

Then Week 68 is the observed abundance peak. The SIMPER results can subsequently tell you whether that peak contributes significantly to differences between Week 68 and other weeks.

That distinction is important:

> **Abundance determines the peak; SIMPER determines whether the abundance difference contributes importantly and consistently to community dissimilarity.**

For each taxon, I would therefore calculate mean abundance by week, locate its maximum, and then check whether comparisons between that peak week and other sampling periods show large and preferably significant SIMPER contributions.

For your Transect A results, this is approximately what led to patterns such as:

**Ceraceomyces → Week 10**
It repeatedly has higher abundance at Week 10 and significant SIMPER contributions in comparisons such as 10–12, 10–14, 10–16, 10–51, 10–52, 10–53, 10–61, and 10–63. 

**Lenzites → Week 12**

**Collophora → Week 14**
It is repeatedly higher around Week 14 and contributes significantly to several contrasts involving that sampling period. 

**Metulodontia → Week 65**
This is particularly clear: mean abundance at Week 65 is 0.455, and it is significant in numerous comparisons involving Week 65. 

**Xenasmatella → Weeks 67–68**
Its abundance becomes very high late in the sampling period—0.945 at Week 67 and 1.000 at Week 68—and it frequently contributes roughly 30–46% of dissimilarity in contrasts with earlier weeks.

I would be more cautious with the sequence I previously described as:

> *Serpula* → *Ceraceomyces* → *Lenzites* → *Collophora* → *Metulodontia* → *Xenasmatella*

That was a useful **pattern-recognition summary**, but it should **not yet be presented as a formal temporal sequence or ecological succession**.

```r
        Max_contribution_pct = max(Contribution_pct),
        Mean_ratio = mean(ratio, na.rm = TRUE),
        .groups = "drop"
    ) %>%
    arrange(
        desc(Significant_contrasts),
        desc(Mean_contribution_pct)
    )
```

This gives you something conceptually like:

| Taxon          | Significant contrasts | Mean contribution | Max contribution |
| -------------- | --------------------: | ----------------: | ---------------: |
| *Xenasmatella* |                  many |               ... |              ... |
| *Metulodontia* |                  many |               ... |              ... |
| *Serpula*      |                  many |               ... |              ... |

I would regard this table as one of the **best ways to identify recurring important contributors**.

### Step 5: Incorporate contribution consistency (`ratio`)

You can make a stricter candidate table:

```r
strong_contributors <- significant_contributors %>%
    filter(ratio >= 1) %>%
    arrange(
        desc(Contribution_pct),
        desc(ratio)
    )
```

I would not make `ratio >= 1` a hard universal statistical threshold. Instead, use it as supporting information:

* higher `Contribution_pct` = greater contribution
* lower `p` = stronger permutation evidence
* higher `ratio` = more consistent contribution

### Step 6: Create an overall contributor summary

A useful final query is:

```r
important_taxa <- simper_70 %>%

        Contrasts = n_distinct(Contrast),

            Contrast[p <= 0.05]

        Mean_contribution = mean(Contribution_pct),
        Max_contribution = max(Contribution_pct),

    arrange(
        desc(Significant_contrasts),
        desc(Max_contribution)
    )
```

This is probably the summary I would use to identify the **taxa worth discussing in Key Findings**.

# Task 2 — Identify taxon-specific abundance peaks

This requires a different workflow.

Your dataframe stores abundance twice:

```text
Group_1   Group_2
ava       avb
```

For example:

```text
Taxon           Group_1 Group_2 ava    avb
Xenasmatella    10      67      0.000  0.945
```

You first need to reconstruct:

```text
Taxon × Week → Mean abundance
```

## Step 1: Extract Group 1 abundances

```r
abundance_group1 <- simper_df %>%
    transmute(
        Taxon,
        Week = Group_1,
        Mean_abundance = ava
    )
```

## Step 2: Extract Group 2 abundances

```r
abundance_group2 <- simper_df %>%
    transmute(
        Taxon,
        Week = Group_2,
        Mean_abundance = avb
    )
```

## Step 3: Combine both sides of every contrast

```r
taxon_week_abundance <- bind_rows(
    abundance_group1,
    abundance_group2

```

Now there will be many duplicate records because, for example, Week 10 appears in many different pairwise comparisons.

That's expected.

## Step 4: Remove duplicated taxon-week abundance observations

If `ava` and `avb` are deterministic means computed from the same underlying data, they should be identical whenever a Taxon × Week combination is repeated.

Start with:

```r
taxon_week_abundance <- taxon_week_abundance %>%
    distinct(
        Taxon,
        Week,
        Mean_abundance
    )
```

I would also verify that each Taxon × Week has only one abundance value:

```r
taxon_week_abundance %>%
    group_by(Taxon, Week) %>%
    summarise(
        n_values = n_distinct(Mean_abundance),
        .groups = "drop"

    filter(n_values > 1)
```

Ideally this returns **zero rows**.

If so, collapse it:

```r
taxon_week_abundance <- taxon_week_abundance %>%
    group_by(Taxon, Week) %>%
    summarise(
        Mean_abundance = first(Mean_abundance),
        .groups = "drop"
    )
```

Now you have your actual abundance-by-time dataset.

```r
taxon_peaks <- taxon_week_abundance %>%
```

# Step 8: Require recurrence before calling a taxon important

I would go one step further and require a taxon to be significant in, say, **at least two contrasts** before emphasizing its peak.

That's not a formal statistical rule; it's a reporting criterion to avoid highlighting isolated results.

```r
recurring_taxa <- significant_contributors %>%
    count(
        Taxon,
        name = "Significant_contrasts"

    filter(Significant_contrasts >= 2)
```

Then:

```r
recurring_taxon_peaks <- taxon_peaks %>%
```

Now your result is much stronger:

# Step 9: Verify that the peak week actually participates in significant contrasts

This is an important final check.

Suppose *Metulodontia* peaks at Week 65. You want to know whether Week 65 is actually involved in its significant SIMPER contrasts.

First create a long representation of the significant contrasts:

```r
significant_peak_support <- significant_contributors %>%
    select(
        Taxon,
        Contrast,
        Group_1,
        Group_2,
        Contribution_pct,
        ratio,
        p
    ) %>%
    pivot_longer(
        cols = c(Group_1, Group_2),
        names_to = "Group_position",
        values_to = "Week"
    )
```

Then join the peak weeks:

```r
peak_validation <- significant_peak_support %>%
    inner_join(
        taxon_peaks %>%
            select(
                Taxon,
                Peak_week = Week
            ),
        by = "Taxon"
    ) %>%
    mutate(
        Involves_peak = Week == Peak_week
    )
```

Finally summarize:

```r
peak_support_summary <- peak_validation %>%
    group_by(Taxon, Peak_week) %>%
    summarise(
        Significant_contrasts = n_distinct(Contrast),
        Peak_related_contrasts =
            n_distinct(Contrast[Involves_peak]),
        Peak_support_pct =
            100 * Peak_related_contrasts /
            Significant_contrasts,
        .groups = "drop"
    ) %>%
    arrange(
        Peak_week,
        desc(Peak_support_pct)
    )
```

This is a very useful metric.

If you get:

```text
Taxon          Peak week  Significant  Peak-related
Metulodontia   65         12           11
```

that is compelling evidence that the Week 65 abundance maximum is closely related to the taxon's contribution to temporal beta diversity.

# The complete logic

I would think about the analysis as this decision chain:

```text
SIMPER results
     │
     ├── Which taxa explain most dissimilarity?
     │       ↓
     │   cumulative contribution ≈ 70%
     │
     ├── Which contributions are statistically supported?
     │       ↓
     │   p ≤ 0.05
     │
     ├── Which taxa recur across contrasts?
     │       ↓
     │   count significant contrasts
     │
     ├── Are their contributions consistent?
     │       ↓
     │   examine Average / SD
     │
     └── Important SIMPER taxa
               │
               ↓
       Reconstruct abundance
       from Group_1/ava and
       Group_2/avb
               │
               ↓
       Mean abundance by
       Taxon × Week
               │
               ↓
       Find maximum abundance
       for each taxon
               │
               ↓
       Order peaks by week
               │
               ↓
       Verify peak weeks occur
       in significant SIMPER
       contrasts
```

### Most important distinction

I would keep these definitions explicit in your notebook:

> **Important SIMPER contributor:** a taxon accounting for a substantial portion of between-group dissimilarity, especially when its contribution is permutation-significant, consistent, and recurrent across contrasts.

> **Taxon-specific peak:** the Week Elapsed at which that taxon has its highest observed mean abundance.

> **Supported temporal peak:** a taxon-specific abundance peak that is also repeatedly involved in significant SIMPER contrasts.

And only the last two steps give you the basis for saying:

> “The data show a temporal sequence of observed abundance peaks.”

I would **not call that sequence ecological succession yet**. A temporal abundance plot plus Indicator Species Analysis or a dedicated temporal model would provide stronger evidence for that interpretation.
