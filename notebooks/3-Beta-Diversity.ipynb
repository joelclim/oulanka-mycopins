## 1. Introduction and Scope

This module presents the analysis of beta diversity for the Oulanka MycoPins project, extending our investigation from local species richness (as detailed in the Alpha Diversity notebook) to the variation in community composition across the Oulanka sampling gradient. By quantifying community turnover, we aim to characterize how fungal assemblages differ between MycoPin sites and to identify the degree of spatial heterogeneity within the study area. This analysis is critical for understanding whether observed patterns of biodiversity are driven by environmental filtering or by dispersal-related processes such as spatial autocorrelation.

The scope of this analysis encompasses the computation of dissimilarity matrices followed by multivariate ordination techniques, specifically Non-metric Multidimensional Scaling (NMDS). We will evaluate the structure of these communities in relation to the environmental covariates identified during our Explorative Data Analysis (EDA) phase within the Oulanka landscape. All computations rely on the standardized abundance matrix generated and validated in the preceding analytical workflows.

## 2. Research Objectives

The primary objective of this analysis is to quantify the magnitude and drivers of fungal species turnover across the Oulanka MycoPins sampling gradient. By characterizing the degree of community divergence between MycoPin sites, we aim to determine whether biodiversity patterns are characterized by high species replacement or by a nested pattern where certain locations represent subsets of others. Ultimately, this work seeks to differentiate between deterministic processes—such as environmental filtering and niche partitioning within boreal microhabitats—and stochastic processes, such as dispersal limitation and priority effects, in shaping the observed community structures.

Specifically, this analysis is designed to address the following research questions:

1.  **Transect-Level Community Differentiation:** Does fungal community composition differ significantly among Transects A, B, and C, and are these patterns consistent between the complete fungal community and the saproxylic fungal subset?
2.  **Within-Transect Community Heterogeneity:** Do Transects A, B, and C differ in the degree of within-transect variation in fungal community composition?
3.  **Substrate and Temporal Structuring:** Within individual transects, are differences in saproxylic fungal community composition associated with wood substrate characteristics (wood texture and wood type) or elapsed sampling time?

## 3. Key Findings
Beta diversity analyses revealed significant differences in fungal community composition among transects for both the complete fungal community and the subset of saproxylic fungi, while analyses within the saproxylic community indicated that temporal variation was more consistently associated with community composition than either wood texture or wood type (Tables 1 and 3; Figures 1 and 3).

### Community differences among transects
When all fungal taxa were considered, community composition differed significantly among Transects A, B, and C (PERMANOVA: _p_ = 0.001, _R²_ = 0.113; ANOSIM: _p_ = 0.001, _R_ = 0.392; Table 1). Transect therefore accounted for approximately 11.3% of the variation in overall fungal community composition. The corresponding NMDS ordination (Figure 1) provides a visual representation of differences in fungal assemblage structure among the three transects.

A similar pattern was observed when the analysis was restricted to saproxylic fungi. Community composition again differed significantly among transects (PERMANOVA: _p_ = 0.001, _R²_ = 0.102; ANOSIM: _p_ = 0.001, _R_ = 0.228; Table 1), with transect accounting for approximately 10.2% of the variation in saproxylic fungal composition.

The magnitude of differentiation among transects was somewhat greater for the complete fungal community. All Fungi produced a slightly higher PERMANOVA _R²_ (0.113 versus 0.102) and a higher ANOSIM _R_ statistic (0.392 versus 0.228) than the saproxylic subset. However, because these analyses contain substantially different numbers of taxa, these statistics should be interpreted as complementary rather than as direct measures of the relative ecological importance of transect.

NMDS representation was also stronger for the saproxylic fungal subset. The saproxylic ordination had a stress value of 0.137, indicating a generally acceptable two-dimensional representation of the underlying community dissimilarities, whereas the All Fungi ordination had a higher stress of 0.202 (Table 1). Fine-scale distances and apparent differences in hull or ellipse geometry in the All Fungi ordination should therefore be interpreted cautiously.

### Differences in community dispersion
Multivariate dispersion differed significantly among transects for both the complete fungal community (betadisper: _p_ = 0.010) and the saproxylic fungal fungi (_p_ = 0.001; Table 1). Tukey-adjusted pairwise comparisons (Table 2) showed that, for All Fungi, Transect C was significantly more dispersed than Transect A, whereas for saproxylic fungi, Transect C was significantly more dispersed than both Transects A and B.

```r
all_vs_sap_tukey_table(tab_n = "2")
```

For All Fungi, Tukey-adjusted pairwise comparisons identified only one significant contrast. Transect C was significantly more dispersed than Transect A (difference in mean distance to centroid = 0.0418, adjusted _p_ = 0.0094). Dispersion did not differ significantly between Transects A and B (adjusted _p_ = 0.396) or between Transects B and C (adjusted _p_ = 0.136). Thus, although Transect B appears to occupy a relatively large area in the two-dimensional NMDS ordination, the formal dispersion analysis does not support greater community heterogeneity in Transect B. Rather, the statistically supported difference is greater dispersion in Transect C relative to Transect A.

For saproxylic fungi, the dispersion pattern was substantially stronger. Transect C was significantly more dispersed than Transect A (difference = 0.1096, adjusted _p_ < 0.001) and Transect B (difference = 0.1408, adjusted _p_ < 0.001), whereas Transects A and B did not differ significantly (adjusted _p_ = 0.334). The greater dispersion of Transect C is also apparent in the saproxylic fungal ordination and indicates that saproxylic fungal communities were more compositionally heterogeneous among samples in Transect C than in either Transect A or B.

A plot of sample distances to their respective group centroids (Figure 2) further illustrates these differences in multivariate dispersion. In particular, the saproxylic fungal community shows a broader distribution and greater average distance to centroid in Transect C, consistent with the overall betadisper test and Tukey-adjusted pairwise comparisons.

Because significant differences in dispersion were detected, the significant transect PERMANOVA results should be interpreted cautiously. PERMANOVA can be influenced by both differences in group centroids and differences in within-group dispersion. The transect effects therefore provide evidence of overall multivariate differentiation among fungal communities, but they cannot be attributed exclusively to shifts in mean community composition.

### Effects of wood substrate within saproxylic communities
Within individual transects, wood texture showed little evidence of structuring saproxylic fungal communities (Table 2). PERMANOVA was not significant in Transects A (_p_ = 0.833, _R²_ = 0.006) or B (_p_ = 0.407, _R²_ = 0.012). In Transect C, the PERMANOVA result approached but did not reach the conventional significance threshold (_p_ = 0.048, _R²_ = 0.034), although ANOSIM detected a weak difference (_p_ = 0.027, _R_ = 0.094). The small effect sizes and disagreement between the two tests suggest that any association between wood texture and community composition in Transect C was weak.

Similarly, wood type was not significantly associated with saproxylic fungal community composition within any transect (Table 3). PERMANOVA results were non-significant for Transects A (_p_ = 0.815, _R²_ = 0.016), B (_p_ = 0.609, _R²_ = 0.020), and C (_p_ = 0.267, _R²_ = 0.053). ANOSIM likewise provided no significant evidence of community separation by wood type. Differences among birch, pine, and spruce therefore did not appear to be major determinants of saproxylic fungal beta diversity within the transects.

### Temporal variation in saproxylic fungal communities
In contrast to the wood substrate variables, Weeks Elapsed was consistently associated with saproxylic fungal community composition within all three transects (Table 3; Figure 3).

```r
#| fig-align: center
suppressMessages(suppressWarnings({
    options(repr.plot.width = 8, repr.plot.height = 12)
    par(mar = c(8, 5, 5, 1),
        xpd = TRUE    # allow drawing outside plot area
    )
    sap_transects_weeks_elapsed_plot(fig_n = "3")
}))
```

In Transect A, Weeks Elapsed explained approximately 9.1% of compositional variation (PERMANOVA: _p_ = 0.001, _R²_ = 0.091; ANOSIM: _p_ = 0.001, _R_ = 0.175). In Transect B, Weeks Elapsed explained approximately 7.5% of variation (PERMANOVA: _p_ = 0.001, _R²_ = 0.075) and produced the strongest ANOSIM separation among the three transects (_p_ = 0.001, _R_ = 0.374). In Transect C, Weeks Elapsed explained approximately 5.0% of compositional variation (PERMANOVA: _p_ = 0.001, _R²_ = 0.050; ANOSIM: _p_ = 0.001, _R_ = 0.241).

The temporal results require different levels of caution among transects. Multivariate dispersion differed significantly among Weeks Elapsed groups in Transects A (betadisper: _p_ = 0.003) and C (_p_ = 0.001), indicating that their significant PERMANOVA results may partly reflect differences in within-group heterogeneity. In contrast, dispersion did not reach statistical significance in Transect B (_p_ = 0.040). The Transect B result therefore provides the clearest evidence that Weeks Elapsed was associated with a shift in fungal community composition rather than primarily with differences in dispersion.

### Overall pattern
Taken together, the statistical analyses (Tables 1 and 3) and ordination and dispersion visualizations (Figures 1–3) indicate that transect and temporal variation were the principal detectable sources of beta-diversity structure, whereas wood texture and wood type showed comparatively little association with saproxylic fungal community composition.

Differences among transects involved not only shifts in overall multivariate community structure but also differences in within-transect heterogeneity. Transect C emerged as particularly heterogeneous, showing significantly greater dispersion than Transect A for the complete fungal community and significantly greater dispersion than both Transects A and B for saproxylic fungi. Within the saproxylic fungal subset, Weeks Elapsed was the most consistent within-transect predictor of compositional variation, with the strongest evidence for a temporal compositional shift occurring in Transect B.

## 4. Computational Workflow

**Input:** 

- $\mathbf{X}$: Species/OTU abundance matrix.
- $\mathbf{E}$: Environmental metadata (continuous and categorical).
- $\mathbf{G}$: Grouping factor (e.g., Treatment vs. Control).

---

### Phase I: Distance & Dimensionality Reduction
1.  **Compute Dissimilarity Matrix ($\mathbf{D}$):** 
    Apply metric (e.g., Bray-Curtis) to $\mathbf{X}$.
    *   *R:* `vegdist(X, method = "bray")`
2.  **Perform Ordination:** 
    Project $\mathbf{D}$ into low-dimensional space via NMDS.
    *   *R:* `metaMDS(X, distance = "bray")`
3.  **Gate 1: Stress Validation:**
    *   **If** $\text{Stress} < 0.2$: **Proceed.**
    *   **If** $\text{Stress} \geq 0.3$: **STOP.** (Ordination is arbitrary; re-evaluate scaling or metric).

---

### Phase II: Primary Hypothesis Testing (The Decision Engine)
4.  **Execute PERMANOVA:** Test if groups ($\mathbf{G}$) drive composition.
    *   *R:* `adonis2(D ~ G, data = E)`
5.  **Gate 2: Significance & Effect Size Evaluation:**
    *   **Check $p$-value ($Pr(>F)$):**
        *   If $p \geq 0.05$: **STOP/CAUTION.** (No statistical difference detected).
        *   If $p < 0.05$: **PROCEED** to Step 6.
    *   **Check $R^2$:**
        *   If $R^2$ is high ($> 0.3$): High explanatory power; proceed to visualization.
        *   If $R^2$ is low ($< 0.1$): Significant but weak effect; proceeding requires caution (overlap will be high).

6.  **Gate 3: The Dispersion Diagnostic (The "Gotcha" Check):** 
    Run `betadisper` to distinguish **Shift** from **Spread**.
    *   *R:* `permutest(betadisper(D, G))`
    *   **Scenario A:** $p_{\text{perm}} > 0.05$ (Equal Dispersion) $\rightarrow$ **Result:** True Compositional Shift. (High confidence).
    *   **Scenario B:** $p_{\text{perm}} < 0.05$ (Unequal Dispersion) $\rightarrow$ **Result:** Variance-driven significance. (Interpret `envfit` with extreme caution; the "difference" may just be increased instability in one group).

*(Optional Parallel Step: Run ANOSIM to validate strength of separation via $R$-statistic.)*

---

### Phase III: Driver Identification & Visualization
7.  **Identify Environmental Drivers (`envfit`):** 
    Correlate continuous/categorical variables from $\mathbf{E}$ with the ordination coordinates.
    *   *R:* `envfit(ordination, E)`
8.  **Final Synthesis (The Plot):** 
    Map the spatial distribution of sites, group centroids, and environmental vectors.
    *   *R:* `ordiplot(ordination)`

---

### Summary Logic Table for Final Interpretation

| PERMANOVA ($p$) | `betadisper` ($p$) | Ecological Conclusion | Action |
| :--- | :--- | :--- | :--- |
| **Significant** | **Not Significant** | **True Compositional Shift** | Proceed to `envfit` & Plotting. |
| **Significant** | **Significant** | **Dispersion/Variance Effect** | Proceed with extreme caution; focus on stability. |
| **Not Significant**| **Not Significant** | **No Difference** | Stop; no pattern to report. |

```r
clean_dataset <- function(community, environment) {
    clean_community <- as.data.frame(community)
    clean_environment <- as.data.frame(environment)

    rownames(clean_community) <- environment$Sample.Number
    rownames(clean_environment) <- environment$Sample.Number
    
    samples_with_zero_abundance <- rownames(clean_community)[rowSums(clean_community) == 0]

    clean_community = clean_community[
        !rownames(clean_community) %in% samples_with_zero_abundance, , drop = FALSE
    ]
    
    missing_samples <- setdiff(
        rownames(clean_community),
        rownames(clean_environment)
    )
    
    if (length(missing_samples) > 0) {
        stop(
            "Samples missing from environment: ",
            paste(missing_samples, collapse = ", ")
        )
    }
    
    clean_environment <- clean_environment[
        rownames(clean_community),
        ,
        drop = FALSE
    ]    
    
    return (list(
        community = clean_community,
        environment = clean_environment
    ))
}
```

```r
run_ordination <- function(configuration) {
    suppressMessages(suppressWarnings({
        ordination = list()
        ordination$configuration <- configuration
        
        # Calculate ecological distance
        ordination$dist <- vegdist(configuration$community, configuration$method)
        # Compute dissimilarities
        ordination$mds <- metaMDS(ordination$dist, 
                                  try = 1000, 
                                  distance = configuration$method, 
                                  k = 2)
        
        # Hypothesis Test (PERMANOVA)
        formula <- reformulate(configuration$variable, response = "ordination$dist")
        ordination$permanova <- adonis2(formula, data = configuration$environment)
    
        # Diagnostic Hypothesis Test (Dispersion)
        ordination$dispersion_model <- betadisper(ordination$dist, 
                                       configuration$environment[[configuration$variable]])
        ordination$dispersion <- permutest(ordination$dispersion_model, 
                                           permutations = 999)
        # Which pairs of group means are statistically significantly different
        ordination$dispersion_tukey_hsd <- TukeyHSD(ordination$dispersion_model)
    
        # Hypothesis Test (ANOSIM)
        ordination$anosim <- anosim(ordination$dist, 
                                    configuration$environment[[configuration$variable]])
        
        # Hypothesis Test results
        ordination$hypothesis_test <- list(
            "permanova.p_value" = ordination$permanova["Model", "Pr(>F)"],
            "permanova.r_squared" = ordination$permanova["Model", "R2"],
            "betadisper.p_value" = ordination$dispersion$tab["Groups", "Pr(>F)"],
            "anosim.p_value" = ordination$anosim$signif,
            "anosim.r_statistic" = ordination$anosim$statistic
        )
        
        # Fit an environmental vector on the ordination plot
        ordination$vector <- envfit(ordination$mds, 
                                    configuration$environment[configuration$vector], 
                                    perm = 1000, 
                                    arrows = TRUE)
    }))
    
    return(ordination)
}
```

```r
nmds_plot <- function(options) {    
    plot <- ordiplot(options$mds, 
                     type = options$graph_type, 
                     xlim = options$xlim, 
                     ylim = options$ylim)
    if ("ellipse" == options$draw) {
      ordiellipse(options$mds, 
                  groups = options$groups,
                  draw = "polygon", 
                  col = options$colors, 
                  label = options$draw_label)
    
    } else if ("hull" == options$draw) {
      ordihull(options$mds, 
               groups = options$groups,
               draw = "polygon", 
               col = options$colors, 
               label = options$draw_label)
    }
    
    if (options$show_sites) {
      orditorp(options$mds, 
               display = "sites", 
               cex = 0.78, 
               air = 0.01)
    }
    if (options$show_species) {
      orditorp(options$mds, 
               display = "species", 
               cex = 0.50, 
               col = "red", 
               air = 0.01)
    }
}
```

```r
all_ordination$transects$nmds_plot <- function(ordination, show_title=TRUE, show_legend=TRUE, fig_n="X") {
    title_text <- "NMDS Ordination of Fungal Communities in Oulanka Forest"
    subtitle_text <- "Multivariate analysis of species composition across varying transects"
    caption_text <- paste0(
        "Figure ", fig_n, ": Non-metric Multidimensional Scaling (NMDS) plot illustrating the separation of fungal communities ",
        "among three experimental transects: Conifer with reindeer access (A), Conifer without reindeer ",
        "access (B), and Broadleaf with tourist access (C)."
    )
    caption_text <- paste(strwrap(caption_text, width = 90), collapse = "\n")

    groups <- ordination$configuration$environment[[ordination$configuration$variable]]
    colors <- viridis(length(unique(groups)))
    
    options <- list(
        mds = ordination$mds,
        groups = groups,        # Transect, Wood.Type, Wood.Texture
        colors = colors,
        vector = ordination$vector,
        graph_type = "none",    # points, text, none
        draw = "ellipse",       # ellipse, hull
        draw_label = FALSE,
        show_sites = FALSE,
        show_species = FALSE,
        xlim = c(-2, 2),        # Zoom in and out
        ylim = c(-2, 2) 
    )
    nmds_plot(options)

    # add environmental vector
    plot(options$vector, arrow.mul = 4, cex=0.90, add=TRUE)

    if (show_title) {
        # add title
        title(main = title_text, col.main = "black", font.main = 2)
        # add caption at th bottom
        title(sub  = caption_text, line = 6, cex.sub = 0.9) 
        # add a tight subtitle right under the main title:
        mtext(subtitle_text, side = 3, line = 0.5, cex = 0.9)
    }

    if (show_legend) {
        # Add the Legend
        groups <- unique(sort(options$groups))
        legend(
          "right",                                     # Position keyword
          legend = groups,                             # Text for items
          col = colors,                                # Colors matching your plot points
          pch = 16,                                    # Point style matching your plot points
          bty = "n",                                   # "n" removes the border box around the legend
          inset = c(-0.40, 0),                         # legend space after plot
          title = ordination$configuration$variable    # Title for the legend itself
        )
    }
}
```

## Beta Diversity Notes

### 1. Calculate Dissimilarities (Creates the Distance Matrix)
* **What you do:** Take your raw community/species abundance data and apply a distance metric (most commonly Bray-Curtis for ecological data).
* **R function**: `vegdist(species_data, method = "bray")`
### 2. Run the Ordination
* **What you do:** Use that distance matrix to project your samples into a low-dimensional space. If you use NMDS, check the stress value here to ensure the math is reliable before moving on.
* **R function:** `metaMDS(species_data, distance = "bray")` (Note: `metaMDS` can take raw data and calculate steps 1 and 2 at the same time).

**General Guidelines in Stress Value Interpretation**
* **< 0.05**: Excellent fit (though values suspiciously close to zero may indicate data problems).
* **< 0.1**: Great or good fit.
* **< 0.2**: Usable or acceptable fit for most community ecology data.
* **> 0.30**: Poor fit; the ordination is close to arbitrary.

### 3. Run Hypothesis Testing (PERMANOVA)
* **What you do:** Statistically test if your predefined groups (e.g., Treatment vs. Control) have significantly different community compositions. You do this next because if the groups aren't significantly different, looking for environmental drivers or plotting them loses its foundation.
* **R function:** adonis2(distance_matrix ~ groups) **(PERMANOVA is widely preferred over ANOSIM in modern ecology)**.
* To determine if you should proceed with envfit and plotting, you need to look at two primary metrics in your PERMANOVA (adonis2) output table: the p-value (Pr(>F)) and the R² value (coefficient of determination).
  1. The P-Value (Pr(>F)): Is there a real pattern?
     * The p-value tells you if the differences in your community structure across groups are statistically significant or just random noise.
       * If **p < 0.05 (Significant)**: Proceed to envfit and plotting. Your groups have distinctly different community compositions. Plotting the ordination will help you visualize these differences, and envfit will help identify which specific environmental variables are driving the shift between these groups.
       * If **p ≥ 0.05 (Not Significant)**: Stop or proceed with caution. Your groups do not statistically differ in community composition. Plotting them will likely just show a messy, overlapping cloud of points, and envfit vectors may not align with any meaningful ecological gradient.
  2. The R² Value: How strong is the pattern?
     * The R² value tells you the effect size—specifically, the proportion of the total variation in your community data that is explained by your grouping variable.
       * **High R² (e.g., > 0.30)**: The grouping variable explains a large amount of the community variance. Your ordination plot will likely show clear, distinct separation between groups.
       * **Low R² (e.g., < 0.10)**: Even if the p-value is significant (which happens easily with large sample sizes), the grouping variable explains very little of the variance. If you plot the ordination, the groups will heavily overlap because 90%+ of the community variation is being driven by other factors.
#### The Crucial "Gotcha": Check for Dispersion (betadisper)
Before making your final decision, you must run a homogeneity of multivariate dispersions test using `betadisper()`. (Note: Checking for dispersion functions as a diagnostic check rather than the primary test.)

PERMANOVA can give a significant p-value for two different reasons:
1. The groups have different averages/centroids (a true directional shift in composition).
2. One group is just way more variable/spread out than the other (dispersion).

**What Hypothesis is betadisper Testing?**
When you use the `betadisper` function in vegan (which implements _Marti Anderson's PERMDISP method_), it tests a specific null hypothesis:
* Null Hypothesis ($H_0$): The average distance of samples to their group centroid (the spread/variance) is equal across all groups.
* Alternative Hypothesis ($H_A$): At least one group has a significantly wider or tighter spread of species composition than the others.

> PERMANOVA (`adonis2`) has a known limitation: it assumes your groups have similar multivariate dispersion (spread). If your groups have unequal spread, PERMANOVA can return a significant p-value even if the groups share the exact same average community composition. Therefore, you run `betadisper` alongside adonis2 to correctly interpret your primary hypothesis test:

If `betadisper` is not significant (p > 0.05), you have equal variance. A significant PERMANOVA means a true shift in community composition, making it highly meaningful to plot and run envfit.

If `betadisper` is significant (p < 0.05), your groups have unequal spread. Your PERMANOVA result might be an artifact of this variance. You can still plot the ordination to see this visually, but interpret your envfit results with extreme caution.

| PERMANOVA Result               | betadisper Result              | What it means for your communities                                                                                                                            |
|--------------------------------|--------------------------------|---------------------------------------------------------------------------------------------------------------------------------------------------------------|
| Significant ((p < 0.05))       | Not Significant ((p > 0.05))   | True Compositional Shift. The communities have physically shifted in ordination space. Excellent candidate for envfit.                                        |
| Significant ((p < 0.05))       | Significant ((p < 0.05))       | Dispersion Effect / Variance. One treatment caused the communities to become highly unpredictable/variable, rather than shifting them in one clean direction. |
| Not Significant ((p ≥ 0.05)) | Not Significant ((p ≥ 0.05)) | No Difference. The communities are statistically identical in both location and spread.                                                                       |
